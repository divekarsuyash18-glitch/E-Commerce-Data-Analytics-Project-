```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = "../Dataset 1/"

customers = pd.read_csv(path + "customers.csv")
products = pd.read_csv(path + "products.csv")
orders = pd.read_csv(path + "orders.csv")
order_items = pd.read_csv(path + "order_items.csv")
payments = pd.read_csv(path + "payments.csv")
shipments = pd.read_csv(path + "shipments.csv")
returns = pd.read_csv(path + "returns.csv")
reviews = pd.read_csv(path + "customer_reviews.csv")
marketing = pd.read_csv(path + "marketing_campaigns.csv")

print("All datasets loaded successfully!")
```

```python
orders['order_date'] = pd.to_datetime(orders['order_date'])
```

```python
## Order Year 

orders['order_date'].dt.year

orders['order_year'] = orders['order_date'].dt.year
```

```python
## Order Month/ month name

orders['order_date'].dt.month

orders['order_month'] = orders['order_date'].dt.month

orders['order_month_name'] = orders['order_date'].dt.month_name()

orders.order_month_name
```

```python
## Order Dayname 

orders['order_date'].dt.day_name()

orders['orders_dayname'] = orders['order_date'].dt.day_name()

orders.orders_dayname
```

```python
## Orders latest Top 5 

orders.head(5)

orders[['order_year', 'order_month', 'order_month_name', 'orders_dayname']].head(5)
```

```python
## Average amount of Orders
orders['final_amount'].mean()

## Median
orders['final_amount'].median()

## highest order value 
orders['final_amount'].max()

## Lowest order value 
orders['final_amount'].min()

## Statical Summary 
orders['final_amount'].describe()

## skewness 
orders['final_amount'].skew()

## finding outliers Q1
orders['final_amount'].quantile(0.25)

## finding outliers Q3
orders['final_amount'].quantile(0.75)

## finding outliers Q2
orders['final_amount'].quantile(0.50)

## IQR
IQR =9503 - 3622
print (IQR)

## Upper limit 
upper_limit = 9503 + (1.5 * 5881)
print (upper_limit)
```

```python
## Exceeding orders on upper limit 

exceeding_orders = orders.sum() > upper_limit
print (exceeding_orders)
```

```python
## Histogram for Order Value Distribution

plt.figure(figsize=(8, 5))
plt.hist(orders['final_amount'], bins=30)
plt.title('Distribution of Order Values')
plt.xlabel('Order Value (₹)')
plt.ylabel('Number of Orders')
plt.show()
```

```python
## Box plot for Order Value Distribution

plt.figure(figsize=(8, 5))
plt.boxplot(orders['final_amount'])
plt.title('Distribution of Order Values')
plt.ylabel('Order Value (₹)')
plt.show()
```

```python
## Calculate total spending per customer
spend_per_customer = orders.groupby('customer_id')['final_amount'].sum()
spend_per_customer.head()

## Average customer spending
avg_spend_customer = spend_per_customer.mean()

## Median customer spending
median_spend_customer = spend_per_customer.median()

## Highest customer spending
max_spend_customer = spend_per_customer.max()

print("Average Customer Spending: ₹", round(avg_spend_customer, 2))
print("Median Customer Spending: ₹", round(median_spend_customer, 2))
print("Highest Customer Spending: ₹", round(max_spend_customer, 2))
```

```python
## Customer spending distribution chart

plt.figure(figsize=(8, 5))
plt.hist(spend_per_customer, bins=30)
plt.title('Customer Spending Distribution Chart')
plt.xlabel('Total Spending per Customer (₹)')
plt.ylabel('Number of Customers')
plt.show()
```

```python
## Top 10% Customer Analysis

top_10_count = int(len(spend_per_customer) * 0.10)

sorted_customers = spend_per_customer.sort_values(ascending=False)

top_10_customers = sorted_customers.head(top_10_count)

top_10_spending = top_10_customers.sum()
total_spending = spend_per_customer.sum()

contribution_percentage = (top_10_spending / total_spending) * 100

print("Top 10% Customer Revenue Contribution:", round(contribution_percentage, 2), "%")
```
